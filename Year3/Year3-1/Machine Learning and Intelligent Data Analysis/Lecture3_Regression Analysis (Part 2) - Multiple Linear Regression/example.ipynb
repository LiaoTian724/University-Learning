# Linear Regression with Scikit-Learn

## 1. Generate Synthetic Data

```python
import numpy as np
import matplotlib.pyplot as plt

# Random x values
x = np.random.rand(50) * 10

# y = 2x + 3 + noise
y = 2 * x + 3 + np.random.randn(50) * 2
```

```python
x.shape
```

```python
y.shape
```

```python
print(x)
print(y)
```

```python
plt.figure(figsize=(4, 4))
plt.scatter(x, y, c="red", ec="black", lw=0.5, label="Observations")
plt.xlabel("x", fontsize=16)
plt.ylabel("y", fontsize=16)
# plt.savefig("example.png", dpi=150, bbox_inches="tight")
```

## 2. Train the Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
# model.fit(x, y)   # this gives error!!

# x reshapes into (n_samples, n_features)

model.fit(x.reshape(-1, 1), y)
```

```python
x
```

```python
y  # note that y will be cast to X's dtype if necesssary
```

```python
# it's a 2-dim numpy array with 1 target and 1 feature
type(model.coef_)

# note numpy.ndarray = n-dimensional array
```

```python
# it's a 1-dim numpy array
type(model.intercept_)
```

```python
print(f"Slope: {model.coef_[0]:.4f}")
print(f"Intercept: {model.intercept_:.4f}")
```

```python
print(model.coef_[0])
print(model.intercept_)
```

## 3. Predict on New Inputs

```python
x_new = np.array([[7], [2], [3]])

pred = model.predict(x_new)

print(pred)
```

```python
print(type(x_new))
print(x_new.shape)
```

```python
plt.figure(figsize=(4, 4))
plt.scatter(x, y, c="red", ec="black", lw=0.5, label="Observations")
plt.scatter(x_new, pred, c="lime", ec="black", lw=0.5, label="New data")
plt.legend(fontsize=12)
plt.xlabel("x", fontsize=16)
plt.ylabel("y", fontsize=16)
# plt.savefig("pred.png", dpi=150, bbox_inches="tight")
```

Seaborn is built directly on top of Matplotlib, serving as a high-level wrapper. It is optimized for fast, beautiful, and complex statistical charts with minimal code.

# Linear Regression by OLS Library 

```python
import statsmodels.api as sm

x = penguins_cleaned[["bill_length_mm"]]
y = penguins_cleaned["flipper_length_mm"]

# Add the intercept term.
x1 = sm.add_constant(x)

# Create and fit the OLS regression model.
olsm = sm.OLS(y, x1)
model = olsm.fit()

# View the fitted coefficients.
model.params
```

```python
model.summary()
```

```python
y_pred = model.predict(x1)

plt.figure(figsize=(3, 3))
plt.scatter(x, y, edgecolor="black", alpha=0.5)
plt.plot(x, y_pred, color="red")
plt.xlabel("bill_length_mm")
plt.ylabel("fipper_length_mm")
plt.title("OLR by statsmodels library")
plt.show()
```

```python
from sklearn.metrics import r2_score, mean_squared_error

slr_r2 = r2_score(y, y_pred)
slr_mse = mean_squared_error(y, y_pred)
slr_rmse = slr_mse**0.5

print(f"SLR R^2: {slr_r2:.4f}")
print(f"SLR MSE: {slr_mse:.2f}")
print(f"SLR RMSE: ${slr_rmse:,.2f}")
```
