```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time

p_head = 0.51
n_tosses = 1000

heads = 0
fractions = []

for i in range(1, n_tosses + 1):

    if np.random.rand() < p_head:
        heads += 1

    fractions.append(heads / i)

    # Update every 20 tosses
    if i % 10 == 0:

        clear_output(wait=True)

        plt.figure(figsize=(8, 5))

        plt.plot(
            range(1, len(fractions) + 1),
            fractions,
            color="blue"
        )

        plt.axhline(
            p_head,
            color="red",
            linestyle="--",
            label=f"True Probability = {p_head:.2f}"
        )

        plt.xlim(1, n_tosses)
        plt.ylim(0, 1)

        plt.xlabel("Number of Tosses", fontsize=16)
        plt.ylabel("Fraction of Heads", fontsize=16)
        plt.title(f"Law of Large Numbers (Tosses = {i})", fontsize=18)

        plt.legend()
        plt.grid(True, alpha=0.3)

        plt.show()

        time.sleep(0.05)
```
