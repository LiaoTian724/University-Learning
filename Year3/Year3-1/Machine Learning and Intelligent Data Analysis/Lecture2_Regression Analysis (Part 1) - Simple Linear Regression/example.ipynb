# Scikit-Learning the Basic of Linear Regression

## 1. Generate Synthetic Data

```python
import numpy as np
import matplotlib.pyplot as plt

# np.random.rand 均匀分布
# np.random.randn 标准正态分布 

# Random x values
x = np.random.rand(50) * 10 

# y = 2x + 3 + noise 
y = 2 * x + 3 + np.random.randn(50) * 2 
```

```python
print(x.shape)
print(y.shape)
```

```python
print(x)
print(y)
```

```python
plt.figure(figsize=(4,4))
plt.scatter(x, y, c="red", ec="black", lw=0.5, label="Observations")
plt.xlabel("x", fontsize=16)
plt.ylabel("y", fontsize=16)
# plt.savefig("example.png", dpi=150, bbox_inches="tight")
```

## 2. Train the Linear Regression Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

x = x.reshape(-1,1)
model.fit(x, y)   # this gives error!! 

# x reshapes into (n_samples, n_features)

# model.fit(x.reshape(-1, 1), y)  
```

```python
x
```

```python
y  # note that y will be cast to X's dtype if necesssary
```

```python
print(x.shape)
print(y.shape)
```

```python
# it's a 2-dim numpy array with 1 target and 1 feature
type(model.coef_)

# note numpy.ndarray = n-dimensional array
```

```python
# it's a 1-dim numpy array
type(model.intercept_)
```

```python
print(f"Slope: {model.coef_[0]:.4f}")
print(f"Intercept: {model.intercept_:.4f}")
```

```python
print(model.coef_[0])
print(model.intercept_)
```

## 3. Predict on New Inputs

```python
x_new = np.array([[7],
                  [2],
                  [3]])

pred = model.predict(x_new)

print(pred)
```

```python
plt.figure(figsize=(4,4))
plt.scatter(x, y, c="red", ec="black", lw=0.5, label="Observations")
plt.scatter(x_new, pred, c="lime", ec="black", lw=0.5, label="New data")
plt.legend(fontsize=12)
plt.xlabel("x", fontsize=16)
plt.ylabel("y", fontsize=16)
# plt.savefig("pred.png", dpi=150, bbox_inches="tight")
```
