### Implementation Tips: 
1. Compute the t-statistic value

2. Define the significance level of 0.05
`alpha = 0.05`

3. Compute degrees of freedom
`dof = n - 2`
Use this when testing a coefficient in a simple linear regression model, where two parameters are estimated: the intercept and slope.

4. Compute the two-tailed p-value 
`2 * t.cdf(-abs(t_stat), dof)`
This selects the lower-tail probability corresponding to the absolute t-statistic and multiplies it by 2 to include the equally extreme tail on the other side of the symmetric t distribution.

5. Compare p-value against the significance level (alpha) to test if there is sufficient evidence to reject H0 at the selected significance level.
